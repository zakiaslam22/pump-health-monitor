### About the sensor labels

The Kaggle dataset hides what each sensor measures. The columns are only named `sensor_00` to `sensor_51`, with no descriptions or units.

**Where the descriptions come from:** IBM used this same public pump dataset in a free training lab for its Maximo maintenance software. That lab lists what IBM believes each sensor measures (for example, `sensor_04` = Motor Speed, `sensor_50` = Pump Discharge Pressure 1).
Source: [IBM Maximo APM lab: Pump Data Dictionary](https://ibm.github.io/maximo-labs-archive/apm_8.7/data_dictionary)

These are IBM's best guesses, not official information from whoever collected the data, so treat them with some caution.

**The reference file:** `reference/ibm_sensor_labels.csv` was put together with help from Claude (an AI assistant). It contains:
- `sensor_column`: the original column name in the data
- `description`: IBM's description of the sensor
- `tag_code`: a short code for the type of measurement (for example VIB = vibration, SPD = speed, PRESS = pressure, TEMP = temperature), added to build readable tag names

sensor_15 is not included, because it was dropped (100% missing) and IBM's list has no entry for it either.

**Tag names:** in an industrial data historian like the AVEVA PI System, every sensor has a tag name that follows a naming convention, usually combining the site, the equipment and the type of measurement. To mimic that, each sensor here gets a PI-style name in the format `PUMP01.<TYPE>.<NUMBER>`, for example `PUMP01.SPD.04`.  The tags use **periods (`.`) instead of underscores**.The tag names reflect the hierarchy, pump, then measurement type, then sensor number, which is a common naming convention. In a real PI system, the full hierarchy would live in Asset Framework. 

```python
tags = labels.copy()
tags.head()
```

```python
tags["tag_name"] = "PUMP01." + tags["tag_code"] + "." + tags["sensor_column"].str[-2:]
tags.head()
```

```python
tags["units"] = "unknown"
tags["missing_pct"] = tags["sensor_column"].map(missing_pct).round(2)
tags.head()
```

```python
def quality_note(pct):
    if pct > 1:
        return f"Flagged: {pct}% missing"
    return "OK"

tags["quality_note"] = tags["missing_pct"].apply(quality_note)
```

```python
tags = tags[["tag_name", "sensor_column", "description", "units", "missing_pct", "quality_note"]]
print(tags.shape)
tags
```
